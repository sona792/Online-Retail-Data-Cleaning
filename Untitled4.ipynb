```python

# ============================================================
# WEEK 2: EXPLORATORY DATA ANALYSIS (EDA)
# Dataset: UCI Online Retail Dataset
# ============================================================


# 1. IMPORT LIBRARIES

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import os
import warnings

warnings.filterwarnings("ignore")

sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")


# ============================================================
# 2. LOAD DATASET
# ============================================================

file_path = r"C:\Users\DELL\Downloads\online+retail\cleaned_online_retail.csv"

df = pd.read_csv(file_path)

print("\nDataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst five records:")
print(df.head())


# ============================================================
# 3. UNDERSTANDING THE DATASET
# ============================================================

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nStatistical Overview:")
print(df.describe())


# ============================================================
# 4. MISSING VALUE ANALYSIS
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())

missing_percentage = (df.isnull().sum() / len(df)) * 100

print("\nMissing Value Percentage:")
print(missing_percentage.round(2))

plt.figure(figsize=(10, 5))
missing_percentage[missing_percentage > 0].plot(kind="bar")
plt.title("Missing Values Percentage by Column")
plt.ylabel("Missing Percentage (%)")
plt.xlabel("Columns")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 5. DESCRIPTIVE STATISTICS
# ============================================================

numeric_columns = ["Quantity", "UnitPrice", "TotalSales"]

print("\nDescriptive Statistics:")
print(df[numeric_columns].describe())

print("\nMedian Values:")
print(df[numeric_columns].median())

print("\nMode Values:")
print(df[numeric_columns].mode().iloc[0])

print("\nVariance:")
print(df[numeric_columns].var())

print("\nStandard Deviation:")
print(df[numeric_columns].std())

print("\nSkewness:")
print(df[numeric_columns].skew())


# ============================================================
# 6. HISTOGRAMS
# ============================================================

plt.figure(figsize=(15, 5))

for i, column in enumerate(numeric_columns, 1):
    plt.subplot(1, 3, i)
    sns.histplot(df[column], bins=50, kde=True)
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)

plt.tight_layout()
plt.show()


# ============================================================
# 7. BOX PLOTS
# ============================================================

plt.figure(figsize=(15, 5))

for i, column in enumerate(numeric_columns, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df[column])
    plt.title(f"Box Plot of {column}")

plt.tight_layout()
plt.show()


# ============================================================
# 8. SCATTER PLOTS
# ============================================================

sample_df = df.sample(min(5000, len(df)), random_state=42)

plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=sample_df,
    x="Quantity",
    y="UnitPrice",
    alpha=0.4
)
plt.title("Relationship Between Quantity and Unit Price")
plt.xlabel("Quantity")
plt.ylabel("Unit Price")
plt.tight_layout()
plt.show()


plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=sample_df,
    x="Quantity",
    y="TotalSales",
    alpha=0.4
)
plt.title("Relationship Between Quantity and Total Sales")
plt.xlabel("Quantity")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()


# ============================================================
# 9. CORRELATION HEATMAP
# ============================================================

correlation = df[numeric_columns].corr()

print("\nCorrelation Matrix:")
print(correlation)

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()


# ============================================================
# 10. COUNTRY-WISE SALES ANALYSIS
# ============================================================

country_sales = (
    df.groupby("Country")["TotalSales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTop 10 Countries by Net Sales:")
print(country_sales.head(10))

plt.figure(figsize=(12, 6))
country_sales.head(10).plot(kind="bar")
plt.title("Top 10 Countries by Net Sales")
plt.xlabel("Country")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 11. MONTHLY SALES TREND
# ============================================================

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

df["YearMonth"] = df["InvoiceDate"].dt.to_period("M")

monthly_sales = (
    df.groupby("YearMonth")["TotalSales"]
    .sum()
)

print("\nMonthly Net Sales:")
print(monthly_sales)

plt.figure(figsize=(12, 6))
monthly_sales.plot(marker="o")
plt.title("Monthly Net Sales Trend")
plt.xlabel("Month")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 12. TOP 10 PRODUCTS BY NET SALES
# ============================================================

product_sales = (
    df.groupby("Description")["TotalSales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTop 10 Products by Net Sales:")
print(product_sales.head(10))

plt.figure(figsize=(12, 6))
product_sales.head(10).sort_values().plot(kind="barh")
plt.title("Top 10 Products by Net Sales")
plt.xlabel("Net Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()


# ============================================================
# 13. TRANSACTION-LEVEL ANALYSIS
# ============================================================

invoice_df = (
    df.groupby(["InvoiceNo", "Country"], as_index=False)
    .agg(
        InvoiceTotal=("TotalSales", "sum"),
        InvoiceDate=("InvoiceDate", "min")
    )
)

print("\nInvoice-Level Summary:")
print(invoice_df["InvoiceTotal"].describe())

plt.figure(figsize=(10, 5))
sns.histplot(
    invoice_df[invoice_df["InvoiceTotal"] > 0]["InvoiceTotal"],
    bins=50,
    kde=True
)
plt.title("Distribution of Positive Invoice Totals")
plt.xlabel("Invoice Total")
plt.tight_layout()
plt.show()


# ============================================================
# 14. HYPOTHESIS TESTING
# ============================================================
# Research question:
# Do UK and non-UK customers have different invoice totals?
#
# H0: There is no difference in invoice totals.
# H1: There is a difference in invoice totals.
#
# Mann-Whitney U test is used because invoice totals
# are generally skewed and need not be normally distributed.


# Exclude cancellation invoices and non-positive totals
invoice_df["InvoiceNo"] = invoice_df["InvoiceNo"].astype(str)

valid_invoices = invoice_df[
    (~invoice_df["InvoiceNo"].str.startswith("C")) &
    (invoice_df["InvoiceTotal"] > 0)
]

uk_sales = valid_invoices.loc[
    valid_invoices["Country"] == "United Kingdom",
    "InvoiceTotal"
]

non_uk_sales = valid_invoices.loc[
    valid_invoices["Country"] != "United Kingdom",
    "InvoiceTotal"
]

print("\nNumber of UK invoices:", len(uk_sales))
print("Number of non-UK invoices:", len(non_uk_sales))

if len(uk_sales) > 0 and len(non_uk_sales) > 0:

    test_result = stats.mannwhitneyu(
        uk_sales,
        non_uk_sales,
        alternative="two-sided"
    )

    print("\nMann-Whitney U Test Results")
    print("U Statistic:", test_result.statistic)
    print("P-value:", test_result.pvalue)

    if test_result.pvalue < 0.05:
        print("Result: Reject the null hypothesis.")
        print("There is statistical evidence of a difference in invoice totals.")
    else:
        print("Result: Fail to reject the null hypothesis.")
        print("There is insufficient statistical evidence of a difference.")

    print("\nMedian UK Invoice Total:", uk_sales.median())
    print("Median Non-UK Invoice Total:", non_uk_sales.median())

else:
    print("Not enough invoice data for hypothesis testing.")


# ============================================================
# 15. ADDITIONAL CUSTOMER ANALYSIS
# ============================================================

customer_sales = (
    df.dropna(subset=["CustomerID"])
    .groupby("CustomerID")["TotalSales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nTop 10 Customers by Net Sales:")
print(customer_sales.head(10))

plt.figure(figsize=(12, 6))
customer_sales.head(10).sort_values().plot(kind="barh")
plt.title("Top 10 Customers by Net Sales")
plt.xlabel("Net Sales")
plt.ylabel("Customer ID")
plt.tight_layout()
plt.show()


# ============================================================
# 16. WEEKDAY ANALYSIS
# ============================================================

df["Weekday"] = df["InvoiceDate"].dt.day_name()

weekday_sales = (
    df.groupby("Weekday")["TotalSales"]
    .sum()
)

weekday_order = [
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
]

weekday_sales = weekday_sales.reindex(weekday_order)

print("\nWeekday Net Sales:")
print(weekday_sales)

plt.figure(figsize=(10, 5))
weekday_sales.plot(kind="bar")
plt.title("Net Sales by Day of the Week")
plt.xlabel("Weekday")
plt.ylabel("Net Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# 17. AUTOMATIC KEY FINDINGS
# ============================================================

print("\n========== KEY FINDINGS ==========")

print("Total Records:", len(df))

print("Number of Countries:", df["Country"].nunique())

print("Number of Unique Products:", df["StockCode"].nunique())

print("Number of Identified Customers:", df["CustomerID"].nunique())

print("Total Net Sales:", round(df["TotalSales"].sum(), 2))

print("Average Line-Level Sales:", round(df["TotalSales"].mean(), 2))

print("Highest Net-Sales Country:", country_sales.idxmax())

print("Highest Net-Sales Product:", product_sales.idxmax())

print("Best Month by Net Sales:", monthly_sales.idxmax())

print("Highest Net-Sales Weekday:", weekday_sales.idxmax())
print(df["Weekday"].value_counts().reindex([
    "Monday", "Tuesday", "Wednesday",
    "Thursday", "Friday", "Saturday", "Sunday"
], fill_value=0))


# ============================================================
# 18. EXPORT ANALYSIS RESULTS
# ============================================================

os.makedirs("EDA_Results", exist_ok=True)

# Export descriptive statistics
df[numeric_columns].describe().to_csv(
    "EDA_Results/descriptive_statistics.csv"
)

# Export country-wise sales
country_sales.to_csv(
    "EDA_Results/country_wise_sales.csv"
)

# Export monthly sales
monthly_sales.to_csv(
    "EDA_Results/monthly_sales.csv"
)

# Export product sales
product_sales.head(20).to_csv(
    "EDA_Results/top_products.csv"
)

# Export customer sales
customer_sales.head(20).to_csv(
    "EDA_Results/top_customers.csv"
)

# Export invoice-level data
invoice_df.to_csv(
    "EDA_Results/invoice_level_analysis.csv",
    index=False
)

print("\nAnalysis results exported successfully!")
print("Check the EDA_Results folder.")


# ============================================================
# 19. FINAL SUMMARY
# ============================================================

print("\n======================================")
print("WEEK 2 EDA COMPLETED SUCCESSFULLY!")
print("======================================")

print("Dataset:", df.shape)
print("Visualizations generated.")
print("Statistical analysis completed.")
print("Hypothesis testing completed.")
print("Analysis results exported.")
```
